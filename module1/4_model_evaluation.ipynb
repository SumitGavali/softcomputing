```python
"""
Phase 1 - Step 4: Validation Plot
------------------------------------
This is the key deliverable plot your brief asks for:
"plot predicted vs actual degradation curve" - proof the model
isn't just guessing randomly.
"""
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

df = pd.read_csv("test_predictions.csv")

# -------------------------------------------------------------
# PLOT A: Scatter of Predicted vs Actual SoH (all test rows)
# A perfect model would put every dot exactly on the diagonal line.
# -------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))

axes[0].scatter(df["SoH"], df["Predicted_SoH"], alpha=0.4, s=12)
axes[0].plot([0, 1], [0, 1], "r--", label="Perfect prediction line")
axes[0].set_xlabel("Actual SoH")
axes[0].set_ylabel("Predicted SoH")
axes[0].set_title("Predicted vs Actual SoH (all unseen test batteries)")
axes[0].legend()
axes[0].set_xlim(0, 1.05)
axes[0].set_ylim(0, 1.05)

# -------------------------------------------------------------
# PLOT B: Degradation curve for ONE example unseen battery
# Shows the model tracing the real shape of decline over cycles -
# this is the "credible" proof the brief is asking for.
# -------------------------------------------------------------
example_battery = df["Battery_ID"].unique()[0]
b = df[df["Battery_ID"] == example_battery].sort_values("Discharge_Index")

axes[1].plot(b["Discharge_Index"], b["SoH"], label="Actual SoH", marker=".", color="black")
axes[1].plot(b["Discharge_Index"], b["Predicted_SoH"], label="Predicted SoH", marker=".", color="orange")
axes[1].set_xlabel("Discharge Cycle Index")
axes[1].set_ylabel("SoH")
axes[1].set_title(f"Degradation Curve: {example_battery} (never seen in training)")
axes[1].legend()

plt.tight_layout()
plt.savefig("validation_plot.png", dpi=130)
print("Saved -> validation_plot.png")
print(f"\nExample battery shown: {example_battery}")
```

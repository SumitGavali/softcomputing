### Stage 2: Extracting the Discharge Capacity
**The Purpose:** The primary goal of analyzing this dataset is to predict the **State of Health (SoH)** and the **Remaining Useful Life (RUL)** of a battery. 

The `Capacity` (measured in Amp-hours) is the definitive measure of a battery's health. 
* As a battery charges and discharges repeatedly, its internal chemistry degrades, and it can hold less charge.
* NASA considered these batteries "dead" when their capacity faded by 30% (from 2.0 Ahr down to roughly 1.4 Ahr).

By extracting the overall `Capacity` for each of the 168 discharge cycles, we can plot a single line graph showing the battery "dying" over time. This flattened trend of `(Cycle Index, Capacity)` is exactly what a Machine Learning model needs to learn how rapidly a battery degrades!

```python
import pandas as pd

# List to hold our flattened data
capacity_data = []

# We'll keep a counter specifically for discharge cycles (1 to 168)
discharge_cycle_number = 1

for i in range(cycles.size):
    cycle = cycles[0, i]
    op_type = cycle['type'][0]
    
    # We only care about discharge cycles because that's when total capacity is evaluated
    if op_type == 'discharge':
        
        # Access the inner data structure
        data = cycle['data']
        
        # In MATLAB structs loaded by scipy.io, the values are often deeply nested arrays.
        # data['Capacity'][0, 0] points to the actual scalar value.
        if 'Capacity' in data.dtype.names:
            capacity_val = data['Capacity'][0, 0][0, 0]
            
            capacity_data.append({
                'Discharge_Cycle': discharge_cycle_number,
                'Capacity_Ahr': capacity_val
            })
            
            discharge_cycle_number += 1

# Convert to a Pandas DataFrame
df_capacity = pd.DataFrame(capacity_data)

# Show the first 5 rows (Fresh Battery) and last 5 rows (Aged Battery)
print("Fresh Battery (First 5 Discharge Cycles):")
display(df_capacity.head())

print("\nAged Battery (Last 5 Discharge Cycles):")
display(df_capacity.tail())
```

3. **Capacity Regeneration (The Spikes):** Do you see those small upwards "spikes" periodically (e.g., around cycle ~60 and ~110)? This is a known phenomenon! A Machine Learning model (like an LSTM or a Time-Series Transformer) would try to use the history to predict not just this continuous line, but also anticipate those little rest spikes to give a highly accurate Remaining Useful Life (RUL) prediction.



### Stage 4: Compare Conditions
Now that we've proved we can easily extract and visualize the degradation of a single battery, we should observe how experimental conditions (from the README files) affect different batteries.

**why I chose those three specific batteries out of the 9 groups:**

1. **B0005**: Found in README_05_06_07_18.txt. Operated at Room Temperature (24°C). Discharged at 2A.
2. **B0032**: Found in README_29_30_31_32.txt. Operated at High Temperature (43°C). Discharged at 4A (or effectively comparable aggressive loads). But mainly, it shows high heat.
3. **B0047**: Found in README_45_46_47_48.txt. Operated at Cold Temperature (4°C). Discharged at 1A (cold temperatures physically prevent high discharge rates).

```python
def extract_capacity(battery_id, temp_label):
    """A helper function to extract purely the capacity degradation of a given battery."""
    mat_file = scipy.io.loadmat(f'{DATA_DIR}/{battery_id}.mat')
    cycles = mat_file[battery_id][0, 0]['cycle']
    
    capacity_data = []
    discharge_cycle_number = 1
    
    for i in range(cycles.size):
        cycle = cycles[0, i]
        if cycle['type'][0] == 'discharge':
            data = cycle['data']
            if 'Capacity' in data.dtype.names:
                # the nested extraction magic
                capacity_val = data['Capacity'][0, 0][0, 0]
                capacity_data.append({
                    'Discharge_Cycle': discharge_cycle_number,
                    'Capacity_Ahr': capacity_val,
                    'Battery_ID': battery_id,
                    'Condition': temp_label
                })
                discharge_cycle_number += 1
                
    return pd.DataFrame(capacity_data)

# Let's extract B0005 (Room Temp), B0032 (Baked at 43C), and B0047 (Frozen at 4C)
df_b0005 = extract_capacity('B0005', '24°C (Room Temp)')
df_b0032 = extract_capacity('B0032', '43°C (Baked)')
df_b0047 = extract_capacity('B0047', '4°C (Frozen)')

# Combine them into one big dataframe
df_comparison = pd.concat([df_b0005, df_b0032, df_b0047], ignore_index=True)

# Plotting them together
plt.figure(figsize=(12, 7))
sns.lineplot(
    data=df_comparison, 
    x='Discharge_Cycle', 
    y='Capacity_Ahr', 
    hue='Condition', # This colors the lines by the temperature condition
    style='Battery_ID',
    markers=True, 
    dashes=False
)

plt.axhline(y=1.4, color='k', linestyle=':', label='End of Life (1.4 Ahr)')
plt.title('Impact of Environmental Temperature on Battery Life')
plt.xlabel('Discharge Cycle Index')
plt.ylabel('Capacity (Amp-hours)')
plt.legend(title='Battery & Temp')
plt.tight_layout()
plt.show()
```

The Scientific Principle: When predicting battery health with Machine Learning, Temperature is the #1 external feature that ruins battery chemistry. To build a good model later, we MUST supply temperature as a training feature.

#### plot analysis

* **The "Room Temp" battery (Blue)**: Lasts roughly 125 cycles before dying (hitting 1.4 Ahr).
* **The "Frozen" battery (Green, 4°C)**: It is an absolute disaster! It starts with barely over 1.0 Ahr capacity off the line. Moving extremely slowly, and it's already dead by NASA standards. Batteries hate being frozen because the internal chemical reactions literally slow down.
* **The "Baked" battery (Orange, 43°C)**: Notice how it starts very high (~1.8 Ahr) but drops far more rapidly and aggressively than the room temp line? Heat increases capacity slightly but destroys the internals fast.
By handpicking these three from the README groups, we have visual proof that we can't just train an ML model blindly. The model must know the Ambient_Temperature to make an accurate prediction.

### Stage 5: Understanding the Time-Series Data (Voltage, Current, Internal Resistance)
To predict the Remaining Useful Life (RUL) properly, we should use the inner curves: `Voltage_measured`, `Current_measured`, `Temperature_measured`, and Internal Resistance. 

As a battery degrades over time:
1. **Voltage Drops Faster**: A degraded battery's voltage curve will collapse much quicker than a healthy one during discharge. 
2. **Internal Resistance Increases**: The dataset actually explicitly records internal resistance in the `impedance` cycles (as `Re` = estimated electrolyte resistance, and `Rct` = charge transfer resistance). As SoH drops, Resistance rises!

Let's visualize exactly what happens to the internal `Voltage Curve` and the internal `Resistance (Re)` of Battery B0005 between when it was brand new vs. when it was dying!

```python
# We will extract the Time vs Voltage series from a FRESH discharge cycle vs an AGED discharge cycle for B0005.

b0005 = scipy.io.loadmat(f'{DATA_DIR}/B0005.mat')['B0005'][0, 0]['cycle']

fresh_voltage = None
fresh_time = None
aged_voltage = None
aged_time = None

discharge_count = 0
for i in range(b0005.size):
    cycle = b0005[0, i]
    if cycle['type'][0] == 'discharge':
        discharge_count += 1
        
        # Grab the 2nd discharge cycle (Fresh)
        if discharge_count == 2:
            fresh_time = cycle['data']['Time'][0,0][0]
            fresh_voltage = cycle['data']['Voltage_measured'][0,0][0]
        
        # Grab the 160th discharge cycle (Aged - almost dead)
        if discharge_count == 160:
            aged_time = cycle['data']['Time'][0,0][0]
            aged_voltage = cycle['data']['Voltage_measured'][0,0][0]

# --- Let's Plot the Voltage Curves ---
plt.figure(figsize=(10, 5))
plt.plot(fresh_time, fresh_voltage, label='Fresh Battery (Discharge Cycle 2)', color='blue')
plt.plot(aged_time, aged_voltage, label='Aged Battery (Discharge Cycle 160)', color='red')

plt.axhline(y=2.7, color='black', linestyle='--', label='Cut-off Voltage (2.7V)')
plt.title('Voltage Collapse: Fresh vs. Aged Battery (B0005)')
plt.xlabel('Time (Seconds)')
plt.ylabel('Voltage (V)')
plt.legend()
plt.tight_layout()
plt.show()
```



### The Time-Series Voltage Collapse (Fresh vs Aged)
Look at the **Blue Curve** (Fresh Battery). It maintained a high voltage (let’s assume powering a laptop screen) efficiently for almost 3,500 seconds before dying. Now look at the **Red Curve** (Aged/Dying Battery). Same battery, same discharge load, but its voltage completely crashes past the 2.7V threshold at under 2,600 seconds! The time it takes for the voltage to collapse is a *massive* predictive feature for Machine Learning!



```python
# Let's extract the Internal Resistance over time from the `impedance` cycles.
# NASA calculated "Re" (electrolyte resistance) from the frequency sweep. 
# As batteries age, Re naturally climbs!

resistance_data = []

# Remember that `b0005` is an array of size 616 containing all operations.
impedance_count = 1
for i in range(b0005.size):
    cycle = b0005[0, i]
    if cycle['type'][0] == 'impedance':
        data = cycle['data']
        # The estimated internal electrolyte resistance
        if 'Re' in data.dtype.names:
            re = data['Re'][0,0][0,0]
            rct = data['Rct'][0,0][0,0]
            
            resistance_data.append({
                'Impedance_Cycle': impedance_count,
                'Re_Ohms': re,
                'Rct_Ohms': rct
            })
            impedance_count += 1
            
# Show the resistance climbing over the battery's lifespan
df_res = pd.DataFrame(resistance_data)

plt.figure(figsize=(10, 5))
sns.lineplot(data=df_res, x='Impedance_Cycle', y='Re_Ohms', color='purple', marker='s')
plt.title('Internal Resistance (Electrolyte) growing over battery lifetime (B0005)')
plt.xlabel('Impedance Check Cycle')
plt.ylabel('Internal Resistance (Ohms)')
plt.tight_layout()
plt.show()
```

###  Internal Electrolyte Resistance (`Re`, from the Impedance Data)
You asked about internal resistance. The dataset tracks this exactly in the `impedance` arrays! NASA calculated the internal impedance using an EIS sweep (sending different radio frequencies through the battery to measure internal resistance).
* Look at the climbing purple line. 
* As the battery aged over its lifetime, internal physical damage meant it was physically harder to push electrons through. **The internal resistance (`Re_Ohms`) literally doubled over its lifespan.** 
* When resistance goes up, the battery creates more wasted heat instead of power, which explains why the "Aged" battery voltage crashes so fast!

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=df_res, x='Impedance_Cycle', y='Rct_Ohms', color='green', marker='s')
plt.title('Internal Resistance (Electrolyte) growing over battery lifetime (B0005)')
plt.xlabel('Impedance Check Cycle')
plt.ylabel('Internal Resistance (Ohms)')
plt.tight_layout()
plt.show()
```

# Feature Engineering

### Stage 6: The Ultimate Feature Engineering Script
Instead of just gathering metadata, we are now going to extract meaningful physics-based features from the dense time-series arrays inside the `discharge` cycles, and fetch the nearest internal resistance from the `impedance` cycles.

```python
import os
import numpy as np
import scipy.io
import pandas as pd

data_dir = DATA_DIR
export_dir = EXPORT_DIR
os.makedirs(export_dir, exist_ok=True)

mat_files = [f for f in os.listdir(data_dir) if f.endswith('.mat')]
mat_files.sort()

all_ml_data = []
print(f"Building Advanced ML Dataset for {len(mat_files)} batteries...")

total_dropped_rows = 0

for mat_file in mat_files:
    battery_id = mat_file.split('.')[0]
    file_path = os.path.join(data_dir, mat_file)
    
    try:
        mat = scipy.io.loadmat(file_path)
        cycles = mat[battery_id][0, 0]['cycle']
        
        discharge_index = 0
        last_known_re = None  
        last_impedance_cycle = -1
        first_capacity = None
        
        for i in range(cycles.size):
            cycle = cycles[0, i]
            op_type = cycle['type'][0]
            temp_ambient = cycle['ambient_temperature'][0, 0]
            data = cycle['data']
            
            # 1. Update Internal Resistance when an impedance cycle happens
            if op_type == 'impedance':
                if 'Re' in data.dtype.names:
                    # Extract the Real component of the complex number
                    last_known_re = float(np.real(data['Re'][0,0][0,0]))
                    last_impedance_cycle = i
            
            # 2. Extract deep features when a discharge cycle happens
            elif op_type == 'discharge':
                discharge_index += 1
                
                capacity_val = None
                if 'Capacity' in data.dtype.names:
                    capacity_val = data['Capacity'][0,0][0,0]
                    
                    if first_capacity is None:
                        first_capacity = capacity_val
                else:
                    total_dropped_rows += 1
                    continue
                
                # Fetch arrays
                times = data['Time'][0,0][0]
                voltages = data['Voltage_measured'][0,0][0]
                temps = data['Temperature_measured'][0,0][0]
                
                # Feature Calculations
                discharge_time_seconds = times[-1] if len(times) > 0 else 0
                max_temp_reached = np.max(temps) if len(temps) > 0 else temp_ambient
                min_voltage_recorded = np.min(voltages) if len(voltages) > 0 else 0
                
                v_drop_rate = 0
                if discharge_time_seconds > 0 and len(voltages) > 0:
                    v_drop_rate = (np.max(voltages) - min_voltage_recorded) / discharge_time_seconds
                
                cycles_since_impedance = (i - last_impedance_cycle) if last_impedance_cycle != -1 else None
                soh = capacity_val / first_capacity if first_capacity else None

                all_ml_data.append({
                    'Battery_ID': battery_id,
                    'Cycle_Index': i + 1,
                    'Discharge_Index': discharge_index,
                    'Ambient_Temperature': temp_ambient,
                    'Discharge_Time_Seconds': discharge_time_seconds,
                    'Max_Temp_Reached': max_temp_reached,
                    'Min_Voltage_Recorded': min_voltage_recorded,
                    'Voltage_Drop_Rate_V_per_sec': v_drop_rate,
                    'Internal_Resistance_Re': last_known_re,
                    'Cycles_Since_Impedance': cycles_since_impedance,
                    'Capacity_Ahr': capacity_val,
                    'SoH': soh
                })
                
    except Exception as e:
        print(f"Skipping {battery_id} due to parse error: {e}")

# Export the engineered dataframe
df_ml_dataset = pd.DataFrame(all_ml_data)

# FIX: Backfill the first few rows where a discharge cycle happened before the very first impedance cycle!
# For Cycles_Since_Impedance, we can fill with 0 to say "measured right now" or keep as median.
df_ml_dataset['Internal_Resistance_Re'] = df_ml_dataset.groupby('Battery_ID')['Internal_Resistance_Re'].bfill()
df_ml_dataset['Cycles_Since_Impedance'] = df_ml_dataset['Cycles_Since_Impedance'].fillna(0)

print("*"*20)
print(df_ml_dataset['Internal_Resistance_Re'].isna().sum(), "rows still have NaN Re")
print("*"*20)

export_path = os.path.join(export_dir, 'ML_Engineered_Battery_Dataset.csv')
df_ml_dataset.to_csv(export_path, index=False)

print("\n--- FEATURE ENGINEERING COMPLETE ---")
print(f"Dataset securely saved to: {export_path}")
print(f"Total valid ML rows generated: {len(df_ml_dataset)}")
print(f"Total dropped rows (missing Capacity): {total_dropped_rows}")
display(df_ml_dataset.head())
```

### 📚 Understanding the Engineered ML Features (Stage 6 Breakdown)
The massive script above transformed 34 highly complex, nested MATLAB struct files into a single, clean `.csv` designed specifically for Machine Learning. Because battery degradation is a time-series problem inside a cycle-series problem, we had to extract single numerical features from arrays of thousands of data points.

Here is a clear reference guide explaining every column we engineered, where it comes from, and why it is mathematically necessary for predicting the battery's Health (SoH) or Remaining Useful Life (RUL):

#### 1. The Metadata
* **`Battery_ID` & `Cycle_Index` & `Discharge_Index`**: These identify the exact battery and its timeline position. Notice we built `Discharge_Index` because a battery might run 600 absolute cycles, but only 160 of them are actually *discharge* operations where testing happens.
* **`Ambient_Temperature`**: The environmental condition (e.g., 4°C, 24°C, 43°C). Extremely important because heat accelerates degradation while cold temporarily suppresses capacity.

#### 2. Features Extracted from the Discharge Curves (The Time-Series Arrays)
During a discharge cycle, the battery sensors record high-resolution arrays of Voltage, Current, and Temperature over Time. We flattened those into singular ML features:
* **`Discharge_Time_Seconds`**: How long (in seconds) the battery was able to output power before hitting its voltage cut-off. _(Why: An aging battery holds less charge, meaning it dies faster. This directly correlates with remaining capacity.)_
* **`Max_Temp_Reached`**: The peak physical temperature of the battery during the load. _(Why: As degradation occurs, internal friction increases. Older batteries generate significantly more waste heat than fresh ones under the exact same load.)_
* **`Min_Voltage_Recorded`**: The absolute lowest voltage before the safety cut-off stopped the test. _(Why: Different NASA groups had different cut-offs like 2.0V vs 2.7V. A model needs to know the cutoff baseline to properly contextualize the total capacity emitted.)_
* **`Voltage_Drop_Rate_V_per_sec`**: Calculated as `(Max Voltage - Min Voltage) / Total Time`. _(Why: Fresh batteries maintain a stable, flat voltage. Degraded batteries suffer from "Voltage Sag" and collapse quickly. This gives the ML model the "slope" of the collapse.)_

#### 3. Features Extracted from the Impedance Cycles
NASA periodically paused the charge/discharge cycles to run an Electromagnetic Impedance Spectroscopy (EIS) sweep.
* **`Internal_Resistance_Re`**: NASA calculated `Re` (Electrolyte Resistance). As the internal chemistry physically corrodes, resistance climbs. _(Fix Applied: NASA stored this as a complex imaginary number like `NaN+0.0j`. We solved this by using `np.real()` to extract only the real floating-point value. We also backfilled the first few rows because discharge cycles often began before the very first impedance check was run.)_
* **`Cycles_Since_Impedance`**: Just because we have an `Re` value doesn't mean it's fresh. Impedance was only checked every few dozens of cycles. _(Why: This tells the ML model exactly how "stale" the Resistance number is, preventing it from incorrectly assuming a 10-cycle-old resistance reading is perfectly accurate for the current cycle.)_

#### 4. The Target Variables (Y)
* **`Capacity_Ahr` (Raw Target)**: The total Amp-hours emitted. _(Why it's flawed: Different battery groups used vastly different definitions of "dead", and started with different absolute capacities. You cannot train one universal model on this.)_
* **`SoH` (State of Health - Normalized Target)**: Calculated dynamically as `Current Capacity / First Recorded Capacity`. _(The Solution: This intelligently maps every single battery to an initial health of `1.0` (100%). It allows us to predict a universal End-of-Life (EoL) threshold—like 0.70 or 70%—regardless of what the battery's absolute Amp-hour rating was on the factory floor!)_

### Stage 8: Creating the RUL (Remaining Useful Life) Dataset
If you want to train a model to directly predict the *integer number of cycles remaining* instead of the SoH percentage, you need an `RUL` column.

This code block loads the clean dataset from disk and engineers the `RUL` target by subtracting the current cycle from the final cycle recorded for each battery. 

**Note:** This dataset is saved as a completely separate file (`ML_Engineered_Battery_Dataset_RUL.csv`) so it does not conflict with your SoH dataset!

```python
import pandas as pd
import os

export_dir = EXPORT_DIR
soh_dataset_path = os.path.join(export_dir, 'ML_Engineered_Battery_Dataset.csv')

# 1. Load the cleanly engineered dataset from disk
df_rul = pd.read_csv(soh_dataset_path)

# 2. Calculate the maximum cycle reached by each battery
max_cycles = df_rul.groupby('Battery_ID')['Cycle_Index'].max().reset_index()
max_cycles.rename(columns={'Cycle_Index': 'Max_Cycle'}, inplace=True)

# 3. Merge back into the main dataset
df_rul = df_rul.merge(max_cycles, on='Battery_ID')

# 4. Create the RUL target column
df_rul['RUL'] = df_rul['Max_Cycle'] - df_rul['Cycle_Index']

# 5. Export to a separate CSV
rul_export_path = os.path.join(export_dir, 'ML_Engineered_Battery_Dataset_RUL.csv')
df_rul.to_csv(rul_export_path, index=False)

print("--- RUL FEATURE ENGINEERING COMPLETE ---")
print(f"Dataset securely saved to: {rul_export_path}")
display(df_rul.head(10))
```
